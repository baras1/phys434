```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

print(stats.norm.cdf(1, loc=0, scale=1))
```

```python
print(f"0.6827 is {(stats.norm.ppf(0.6827))} sigma")
print(f"{int(1-0.9545)} is {(stats.norm.ppf(1-0.9545))} sigma")
#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
d = stats.rayleigh.rvs(loc = 0.0, scale = .01, size = 100000)

fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([-0.01,0.05])
x = np.linspace(-0.01,0.05,1000)
ax.plot(x,stats.rayleigh.pdf(x,loc = 0., scale = 0.01),linewidth = 8,alpha = 0.7)
plt.show()

question = "What is the probability that background noise would occur at a value above 0.03?"

p = stats.rayleigh.cdf(0.03, loc=0, scale=.01)
sigma_value = stats.norm.ppf(p)
answer = f"\nThe probability that background noise would occur at a value above 0.03 is {sigma_value} sigma"
print(question, answer)
```
