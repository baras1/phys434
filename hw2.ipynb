```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#Problem 1: A little introductory brain teaser. Which is more probable when rolling 2 six-sided dice: 
#rolling snake eyes (two ones) or rolling sevens (dice sum to seven)? What is the ratio of the probabilities?

p_snake_eyes = (1/6) * (1/6)
p_sum_7 = (1/6) * (1/6) * 6
ratio_snakeeyes_to_sum7 = p_snake_eyes / p_sum_7

print("Problem 1:")
print(f"Probability of Snake Eyes: {p_snake_eyes*100}%")
print(f"Probability of a sum of 7: {p_sum_7*100}%")
print(f"Ratio: {ratio_snakeeyes_to_sum7}")

#Problem 2: Following what we did in class show how to use the convolution operator to determine the probability of the sum of 2 six sided dice. 
#Do both analytically (math & counting) and numerically (computer program). Beware the implicit definition of the values (x-axis on histogram).

die1 = np.ones(6) / 6
die2 = np.ones(6) / 6

sum_probabilities = np.convolve(die1, die2, mode='full')

sums = np.arange(2, 13)

print("\n\nProblem 2:")
print("These are the probabilities associated with each 2-sided die sum")
for s, p in zip(sums, sum_probabilities):
    print(f"Sum = {s}, Probability = {p}")

#Problem 3: Calculate the mean and the variance of the distribution in problem 2. Hint: this is surprisingly tricky, make sure your result makes sense.

print("\n\nProblem 3:")
mean = np.sum(sums * sum_probabilities)
print(f"The mean is {mean}")
variance = np.sum((sums - mean)**2 * sum_probabilities)
print("The variance is", variance)
#Problem 4: Repeat 2, and graph the average of 10 dice. Is this is a Gaussian distribution? Explain in depth.


print("\n\nProblem 4:\nLooks like a Gaussian distribution.\n")
plt.bar(sums, sum_probabilities)

plt.xlabel("Sum of Two Dice")
plt.ylabel("Probability")
plt.title("Probability Distribution of Two Dice")

plt.show()

#Problem 5: Show that the sum and average of an initially Gaussian distribution is also a Guassian (can be analytic or numerical). 
#How does the standard deviation of the resulting sum or average Gaussian change? This is a hugely important result. 
#Explore what this means for integrating a signal over time. How does this explain error analysis and error propagation you have done in previous classes (classes that assumed Gaussian errors)?
```
